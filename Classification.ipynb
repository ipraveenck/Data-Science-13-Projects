```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target)
print("Missing values:", X.isnull().sum().sum())
```

Missing value: To ensure the dataset has no missing values 

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

Standardizing the dataset and ensuring all the feature are contributing equal data

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

models = {
    "Logistic Regression": LogisticRegression(max_iter=500),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(),
    "k-NN": KNeighborsClassifier()
    }

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    print(f"{name} Accuracy: {acc:.4f}")
    print(classification_report(y_test, y_pred))

print("Model Comparison:")
for model, acc in results.items():
    print(f"{model}: {acc:.4f}")

best_model = max(results, key=results.get)
worst_model = min(results, key=results.get)

print(f"Best Performing Model: {best_model}")
print(f"Worst Performing Model: {worst_model}")
```

Logistic Regression - Works well for binary classification (malignant vs. benign).

Decision Tree - Splits data based on feature thresholds.

Random Forest - Ensemble of decision trees.

Support Vector Machine (SVM) - Finds optimal hyperplane separating classes.

k-Nearest Neighbors (k-NN) - Classifies based on majority vote of nearest neighbors.

Random Forest and SVM perform best on this dataset due to their ability to handle complex boundaries.

Decision Tree often performs worst because of overfitting on small datasets.

Logistic Regression and k-NN usually give solid but slightly lower accuracy compared to ensemble methods.
