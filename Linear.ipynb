```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('winequality-red.csv', sep=';')
df.fillna(df.mean(), inplace=True)
print(df.head())
```

```python
alcohol = df['alcohol'].values
citric_acid = df['citric acid'].values
print(alcohol)
print(citric_acid)
```

```python
X=df[['alcohol', 'density']].values
cov_matrix = np.cov(X.T)
print(cov_matrix)
```

```python
eigen_values,eigen_vectors = np.linalg.eig(cov_matrix)
print("Eigenvalues:", eigen_values)
print("Eigenvectors:", eigen_vectors)
```

```python
sorted_indices = np.argsort(eigen_values)[::-1]
top_eigenvalues = eigen_values[sorted_indices[:2]]
top_eigenvectors = eigen_vectors[:, sorted_indices[:2]]
print("Top Eigenvalues:", top_eigenvalues)
print("Top Eigenvectors:", top_eigenvectors)
```
