```python
import numpy as np
import pandas as pd
df = pd.read_excel("ABC Company.xlsx")
df['Height'] = np.random.randint(150, 181, size=len(df))
print(df)
```

```python
team_counts = df['Team'].value_counts()
team_percentage = (team_counts / len(df)) * 100
print(team_counts)
print(team_percentage)
```

```python
employee_seg=df['Position'].value_counts()
print(employee_seg)
```

```python
age_group_counts= df['Age'].value_counts()
predominant_age= age_group_counts.idxmax()
print(predominant_age)
```

```python
salary= df.groupby(['Team','Position'])['Salary'].sum().reset_index()
high_salary= salary.loc[salary['Salary'].idxmax()]
print(high_salary)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
correlation = df[['Age','Salary']].corr()
print(correlation)

plt.figure(figsize=(8,6))
sns.catplot(data=df, x="Age", y="Salary")
plt.title("Age/Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
df['Team'].value_counts().plot(kind='bar', figsize=(8,5))
plt.title("team_counts/team_percentage")
plt.xlabel("team_counts")
plt.ylabel("team_percentage")
plt.show()
```

```python
df['Position'].value_counts().plot(kind='bar', figsize=(8,5))
plt.title("Employee Position")
plt.xlabel("Position")
plt.ylabel("No.of Employee")
plt.show()
```

```python
df['Age'].value_counts().plot(kind='bar', figsize=(8,5))
plt.title("Employee predominant_age")
plt.xlabel("predominant_age")
plt.ylabel("Total predominant_age")
plt.show()
```

```python
salary1= df.groupby(['Team','Position'])['Salary'].sum().reset_index()
high_salary1=salary1.pivot(index='Team', columns='Position', values='Salary').fillna(0)

high_salary1.plot(kind='bar', stacked=True, figsize=(8,5))
plt.title("Salary Expenditure by Team and Position")
plt.xlabel("Team")
plt.ylabel("Total Salary")
plt.legend(title="Position")
plt.show()
```

```python
correlation = df[['Age','Salary']].corr()
plt.figure(figsize=(8,5))
sns.catplot(data=df, x="Age", y="Salary")
plt.title("Age/Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
